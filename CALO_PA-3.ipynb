# Problem A


```python
import pandas as pd 
```

```python
cars = pd.read_csv('cars.csv')
```

```python
#displays the shape of cars
print('Cars:\n', cars.size) 
print('Column Names:\n', cars.columns.tolist()) #display list of columns

#display cars from row6 to row 10
cars_6_to_10 = cars.iloc[5:10]
print('Cars from row 6 to row 10:\n', cars_6_to_10)

#filters list from row6 to row10
filtering = cars_6_to_10[['Model', 'mpg', 'cyl', 'hp', 'gear']]
print('Initialized column names:\n', filtering)
```

# Problem B

```python
toyota = cars.loc[(cars['Model'] == 'Toyota Corolla'), :] #shows the row for toyota corolla 
print('Toyota Corolla Row:\n', toyota)

#displays only selected columns
pontiac = cars.loc[(cars['Model']=='Pontiac Firebird'), ['Model', 'mpg', 'hp', 'wt']]
print('Pontiac Row:\n', pontiac) 
```
